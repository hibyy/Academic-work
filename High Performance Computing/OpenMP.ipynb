```python
%%writefile atomic.c
#include <omp.h>
#include <stdio.h>

int main() {
    int somme = 0;

    #pragma omp parallel for
    for(int i = 0; i < 100000; i++) {
        #pragma omp atomic  //#pragma omp critical
        somme += 1;
    }

    printf("Somme finale  = %d\n", somme);
    return 0;
}
```

```python
%%writefile over.c
#include <stdio.h>
#include <pthread.h>

void* travail(void* arg) {
    double M[2000][2000];

    for(int i=0;i<2000;i++)
        for(int j=0;j<2000;j++)
            M[i][j] = i+j;

    printf("Done\n");
    return NULL;
}

int main() {
    pthread_t tid;

    pthread_create(&tid, NULL, travail, NULL);
    pthread_join(tid, NULL);

    return 0;
}
```
