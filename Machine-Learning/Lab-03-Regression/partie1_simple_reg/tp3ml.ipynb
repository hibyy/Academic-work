```python
from sklearn.model_selection import train_test_split
X= df["Hours"]
Y= df["Scores"]
x_train,x_test, y_train, y_test = train_test_split (X,Y,test_size =0.3,random_state=42)
print("Training samples:", x_train.shape)
print("Testing samples:", x_test.shape)
print(train_test_split(X,Y,test_size =0.3,random_state=42))
```

```python
from sklearn.linear_model import LinearRegression 
model = LinearRegression()
model.fit(x_train.values.reshape(-1, 1), y_train)
```

```python

print("Intercept (b0):", model.intercept_)
print("Slope (b1):", model.coef_)
b0 =  model.intercept_
b1 = model.coef_[0]  
print(f"Equation: Y = {b0:.2f} + {b1:.2f} * X")
```

```python

plt.scatter(df["Hours"], df["Scores"], color='blue', edgecolors='black', label='Data points')
plt.plot(x_train, model.predict(x_train.values.reshape(-1, 1)), color='red', label='Regression line')

plt.title("Linear Regression: Hours vs Scores")
plt.xlabel("Hours studied")
plt.ylabel("Scores obtained")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_train_pred = model.predict(x_train.values.reshape(-1, 1))
y_test_pred = model.predict(x_test.values.reshape(-1, 1))


mae_train = mean_absolute_error(y_train, y_train_pred)
mse_train = mean_squared_error(y_train, y_train_pred)
rmse_train = np.sqrt(mse_train)
r2_train = r2_score(y_train, y_train_pred)

mae_test = mean_absolute_error(y_test, y_test_pred)
mse_test = mean_squared_error(y_test, y_test_pred)
rmse_test = np.sqrt(mse_test)
r2_test = r2_score(y_test, y_test_pred)


print("🔹 Training Set:")
print(f"MAE  = {mae_train:.2f}")
print(f"MSE  = {mse_train:.2f}")
print(f"RMSE = {rmse_train:.2f}")
print(f"R²   = {r2_train:.2f}\n")

print("🔹 Test Set:")
print(f"MAE  = {mae_test:.2f}")
print(f"MSE  = {mse_test:.2f}")
print(f"RMSE = {rmse_test:.2f}")
print(f"R²   = {r2_test:.2f}")
```




**R-squared (R²)** measures how much of the variation in the target variable is explained by the model.
<div>
<ul>
  <li><b>R² = 1</b> → Perfect fit (the model explains 100% of the variance).</li>
  <li><b>R² = 0</b> → The model does not explain any of the variance.</li>
</ul>

<hr>
</div>




The model does **not present an overfitting problem**, because the performance on the training set and the test set is very similar (R² = 0.95 for both). This means the model **generalizes well and 
demonstrates good performance and is not overfitting.**



 **Advantages of R²:**

* It is **easy to interpret** and gives a clear idea of model accuracy.
* It directly shows how well the model **explains the data**.
* Unlike MAE, MSE, or RMSE, R² is **scale-independent**, making it easier to compare models.
