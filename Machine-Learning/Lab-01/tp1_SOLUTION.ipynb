```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

years_of_experience = np.random.randint(1, 11, size=40)

salaries = 2000 + 200 * years_of_experience + np.random.randint(1, 401, size=40)

data = {
    'Years': years_of_experience,
    'Salary': salaries
}

df = pd.DataFrame(data)
print(df) 

df.to_csv('employee.csv' , index=False)

df_loaded = pd.read_csv('employee.csv')
print(df_loaded.head()) 


plt.scatter(df_loaded['Years'], df_loaded['Salary'], color='blue')

plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Salary vs. Years of Experience')

plt.grid(True)
plt.show()



predicted_salary = 2000 + 200 * 3.5 + np.mean([1, 400]) 
print(f"Predicted salary for 3.5 years of experience: {predicted_salary}")
```
