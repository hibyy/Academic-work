```python
lower_bound = Q1 - 1.5 * IQR
print(lower_bound)
upper_bound = Q3 + 1.5 * IQR
print(upper_bound)
data_capped = numeric_data.clip(lower=lower_bound, upper=upper_bound, axis=1)
print(data_capped)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data['class'] = data['class'].map({'live': 1, 'die': 0})

corr = data.corr(method='pearson')
plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Pearson Correlation Heatmap")
plt.show()
```

Pearson correlation measures how linearly related two numeric variables are.
Value range:

+1 → perfectly positive relation

-1 → perfectly negative relation

0 → no linear relation

We use a heatmap to visualize which features are most related to the target variable (class: live/die).

Selected features: albumin


```python
malaise_corr = corr['malaise'].drop('malaise')
high_corr_with_malaise = malaise_corr[abs(malaise_corr) > 0.5].index
print("Highly correlated with malaise:", list(high_corr_with_malaise))

data= data.drop(columns=high_corr_with_malaise)
```

If two input features are highly correlated with each other, they are redundant (collinearity).

## Data splitting

<div class="alert alert-block alert-success">
Data splitting is the process of dividing a dataset into two or more subsets, typically comprising a training set and a testing set. This ensures that models are trained on one subset of the data (training set) while being evaluated on another (test set), which helps effectively assess their performance and improve generalization. In Python, the scikit-learn library is a popular choice for implementing data splitting, offering various functions and classes to facilitate this essential process.
<div/>

<div class="alert alert-block alert-danger">

1. Split the data into training and test sets using the hold-out and stratified hold-out strategies.
2. Display the number of samples in each class for both strategies in the training and test sets.
<div/>

hold_out :Randomly split data into train/test (80% / 20%)
Stratified Hold-out: Ensures both sets have the same class proportions....When you have imbalanced classes


hold out :

```python
import pandas as pd
from sklearn.model_selection import train_test_split

data = pd.read_csv('hepatitis.csv')

data['class'] = data['class'].map({'live': 1, 'die': 0})

X = data.drop(columns=['class'])
y = data['class']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=38
)

print("Hold-out strategy:")
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Class distribution in train:\n", y_train.value_counts())
print("Class distribution in test:\n", y_test.value_counts())
```

satisfied hold-out split :

Both sets have the same class ratio...roughly 80% “live” and 20% “die”.
That’s exactly what we want for fair model evaluation.


```python
import pandas as pd
from sklearn.model_selection import train_test_split

data = pd.read_csv('hepatitis.csv')

data['class'] = data['class'].map({'live': 1, 'die': 0})

X = data.drop(columns=['class'])
y = data['class']

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=38
)

print("Stratified Hold-out Split Done Successfully!")
print("Train class distribution:\n", y_train_s.value_counts())
print("Test class distribution:\n", y_test_s.value_counts())
```
