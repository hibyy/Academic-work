## Répondre aux requêtes MongoDB suivantes :

```python
# Trouver tous les livres empruntés (au moins un exemplaire non disponible)

list(db.livres.find(
    {"exemplaires.status": {"$ne": "emprunte"}},
    {"_id": 1, "title": 1}
))
```

```python
# Trouver tous les livres écrits par l'auteur "Jane Smith"

list(db.livres.find(
    {"authors": "Jane Smith"},
    {"_id": 1, "title": 1}
))
```

```python
# Trouver les livres d’une catégorie donnée (ex: Databases)

list(db.livres.find(
    {"categories": "Databases"},
    {"_id": 1, "title": 1}
))
```

```python
# Trouver les livres publiés après 2019 ET qui ont une catégorie “Programming”

list(db.livres.find(
    {
        "publishedYear": {"$gt": 2019},
        "categories": "Programming"
    },
    {"_id": 1, "title": 1}
))
```

```python
# Livres publiés après 2015 ET appartenant à la catégorie "Databases" (insensible à la casse)

list(db.livres.find(
    {
        "publishedYear": {"$gt": 2015},
        "categories": {"$regex": "^databases$", "$options": "i"}
    },
    {"_id": 1, "title": 1}
))
```

```python
# Nombre total d’emprunts par livre (regrouper par idLivre)

list(db.emprunteurs.aggregate([
    { "$unwind": "$Emprunts" },
    { "$group": {
        "_id": "$Emprunts.idLivre",
        "totalEmprunts": { "$sum": 1 }
    }},
    { "$sort": { "_id": 1 } }
]))
```

```python
#  Trouver les exemplaires disponibles du livre « Learning MongoDB »

list(db.livres.aggregate([
    { "$match": { "title": "Learning MongoDB" }},
    { "$unwind": "$exemplaires" },
    { "$match": { "exemplaires.status": "disponible" }},
    { "$project": { "_id": 0, "exemplaires.copyId": 1 }}
]))
```

```python
#  Trouver les titres des livres empruntés par l’utilisateur « Amine »

list(db.emprunteurs.aggregate([
    { "$match": { "name": "Amine" }},
    { "$unwind": "$Emprunts" },
    {
        "$lookup": {
            "from": "livres",
            "localField": "Emprunts.idLivre",
            "foreignField": "_id",
            "as": "livreInfo"
        }
    },
    { "$unwind": "$livreInfo" },
    { "$project": { "_id": 0, "title": "$livreInfo.title" }}
]))
```

```python
# Lister les livres qui ont au moins deux exemplaires disponibles

list(db.livres.aggregate([
    { "$unwind": "$exemplaires" },
    { "$match": { "exemplaires.status": "disponible" }},
    { "$group": {
        "_id": { "_id": "$_id", "title": "$title" },
        "nbDispo": { "$sum": 1 }
    }},
    { "$match": { "nbDispo": { "$gte": 2 }}}
]))
```
